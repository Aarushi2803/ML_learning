# Feature Engineering

```python
# Removing unncessary columns
data_model = df.copy()

# Drop unwanted columns
data_model.drop(['Unnamed: 0', 'car_name', 'brand', 'model'], axis=1, inplace=True)
data_model.head()
```

```python
# Converting categorical colms into numercial colms
# One hot encoding (function get_dummines)
data_model = pd.get_dummies(data_model,dtype=int) 
data_model 
```

```python
# model creation
# 1. Seprating Indepedent varibales and Depedent varibales

# IVs data
x = data_model.drop(labels='selling_price',axis=1)

# DVs data                    
y = data_model['selling_price']
```

```python
# 2. Train-Test split 
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=0
) 
```

```python
# Model Building
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Train
regressor = LinearRegression()
regressor.fit(x_train, y_train)

# Predict
y_pred = regressor.predict(x_test)

# Evaluation
print("MAE:", mean_absolute_error(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R² Score:", r2_score(y_test, y_pred))

# Adjusted R2 score
# X_test.shape
# r2 = r2_score(y_test,y_pred)
# print("Adjusted r2_score",1 - ((1-r2)*(40-1)/(40-1-1)))  
                                 
```

```python
import pandas as pd
import numpy as np
data = pd.read_csv('housing.csv')
data.head() 

data.drop(columns=['Address','Price'],axis=1,inplace=True)  
data.head() 
```

```python
# Let's apply absolute maximum scaling 
abs_maximum = np.max(np.abs(data),axis=0)
data = data/abs_maximum
data.head() 
# abs_maximum
```

```python
# min-max scaling
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
sacled_data = scaler.fit_tranfrom(data) 
data = pd.DataFrame(scaled_data,columns=data.columns)
data.head() 
```
